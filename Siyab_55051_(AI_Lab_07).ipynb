#**TIC TAC TOE — CLI BASED**

```python
board = ["_" for i in range(9)]

def print_board():
    for i in range(0, 9, 3):
        print(board[i], board[i+1], board[i+2])
    print()

def check_winner():
    win_positions = [
        (0,1,2),(3,4,5),(6,7,8),
        (0,3,6),(1,4,7),(2,5,8),
        (0,4,8),(2,4,6)
    ]
    for pos in win_positions:
        a,b,c = pos
        if board[a] == board[b] == board[c] != "_":
            if board[a] == "x":
                return 10
            else:
                return -10
    return 0

def is_draw():
    return "_" not in board

player = "x"
while True:
    print_board()
    move = int(input(f"Player {player} choose position (1-9): ")) - 1
    if board[move] == "_":
        board[move] = player
    else:
        print("Invalid move!")
        continue
    score = check_winner()
    if score == 10:
        print_board()
        print("Player X wins! Score = +10")
        break
    elif score == -10:
        print_board()
        print("Player O wins! Score = -10")
        break
    elif is_draw():
        print_board()
        print("Game Draw! Score = 0")
        break
    player = "o" if player == "x" else "x"
```

## **TIC TAC TOE — GUI BASED**

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

board = [""] * 9
player = "X"
buttons = [widgets.Button(description="", layout=widgets.Layout(width="60px", height="60px")) for _ in range(9)]
output = widgets.Output()

def check_winner():
    wins = [(0,1,2),(3,4,5),(6,7,8),
            (0,3,6),(1,4,7),(2,5,8),
            (0,4,8),(2,4,6)]
    for a,b,c in wins:
        if board[a] == board[b] == board[c] != "":
            return board[a]
    return None

def on_click(b):
    global player
    i = buttons.index(b)
    if board[i] == "":
        board[i] = player
        b.description = player
        winner = check_winner()
        if winner:
            with output:
                clear_output()
                score = "+10" if winner == "X" else "-10"
                print(f"Player {winner} wins! Score = {score}")
            disable_buttons()
            return
        if "" not in board:
            with output:
                clear_output()
                print("Game Draw! Score = 0")
            return
        player = "O" if player == "X" else "X"

def disable_buttons():
    for b in buttons:
        b.disabled = True

for b in buttons:
    b.on_click(on_click)

grid = widgets.GridBox(buttons, layout=widgets.Layout(grid_template_columns="repeat(3, 60px)"))
display(grid, output)
```
